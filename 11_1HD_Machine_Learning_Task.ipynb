```python
# CONTINUATION AFTER SUCCESSFUL DATASET DOWNLOAD
# HD MINI RESEARCH — OPTION 3

# PART 1:
# Reproducing published-style heart disease classification experiment

# PART 2:
# Investigating duplicate-data leakage + develop a more robust model


# 2. INSTALL / IMPORT REQUIRED LIBRARIES

!pip -q install xgboost

import os
import json
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate,
    RandomizedSearchCV,
    cross_val_score
)

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    StackingClassifier
)

from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    roc_curve,
    auc
)

from xgboost import XGBClassifier


SEED = 42

random.seed(SEED)
np.random.seed(SEED)

OUTPUT_DIR = "/content/HD_Heart_Attack_Research"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


print("CONTINUING HEART DISEASE ML EXPERIMENT")

print("Dataset already loaded:", df.shape)
print("Output folder:", OUTPUT_DIR)


# 3. VERIFY DATASET AGAIN

print("DATASET VERIFICATION")

print("\nDataset shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nTarget distribution:")
print(df["target"].value_counts())

print("\nTarget percentages:")
print(
    df["target"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nMissing values:")
print(df.isnull().sum())

print("\nExact duplicate rows:")
print(df.duplicated().sum())

print("\nUnique rows:")
print(len(df.drop_duplicates()))


# 4. DATASET SUMMARY

summary = df.describe().T

print("DESCRIPTIVE STATISTICS")

display(summary)

summary.to_csv(
    f"{OUTPUT_DIR}/descriptive_statistics.csv"
)


# 5. CLASS DISTRIBUTION

class_counts = (
    df["target"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(7, 5))

plt.bar(
    ["No Heart Disease", "Heart Disease"],
    class_counts.values
)

plt.ylabel("Number of observations")
plt.xlabel("Target Class")

plt.title(
    "Distribution of Heart Disease Classes"
)

for i, value in enumerate(class_counts.values):

    plt.text(
        i,
        value + 5,
        str(value),
        ha="center"
    )

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/01_class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# 6. CORRELATION MATRIX

corr = df.corr(numeric_only=True)

plt.figure(figsize=(12, 9))

plt.imshow(
    corr,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title(
    "Correlation Matrix of Heart Disease Features"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/02_correlation_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# 7. CORRELATION WITH TARGET

target_corr = (
    corr["target"]
    .drop("target")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("FEATURE CORRELATION WITH TARGET")

print(target_corr)

target_corr.to_csv(
    f"{OUTPUT_DIR}/target_correlations.csv"
)


plt.figure(figsize=(10, 6))

plt.barh(
    target_corr.index[::-1],
    target_corr.values[::-1]
)

plt.xlabel(
    "Correlation with Target"
)

plt.title(
    "Feature Correlation with Heart Disease"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/03_target_correlations.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# 8. PREPARE X AND y

X = df.drop(
    columns=["target"]
)

y = df["target"].astype(int)


print("FEATURE MATRIX")

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nPredictor variables:")

for number, feature in enumerate(
    X.columns,
    start=1
):

    print(
        f"{number:2d}. {feature}"
    )


# 9. PART 1 TRAIN / TEST SPLIT

# REPRODUCTION ASSUMPTION:
# We use an 80:20 stratified split and random_state = 42.

# If the paper does not provide an exact executable random seed,
# this is an explicit reproducibility assumption rather than a claim
# that random_state = 42 was used by the original authors.


# For Part 1 we intentionally DO NOT remove duplicates.
# This allows us to evaluate the commonly distributed 1025-row
# dataset in its original form.

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=SEED,

    stratify=y
)


print("PART 1 — TRAIN / TEST SPLIT")

print(
    "Total observations:",
    len(df)
)

print(
    "Training observations:",
    len(X_train)
)

print(
    "Testing observations:",
    len(X_test)
)

print("\nTraining target distribution:")

print(
    y_train.value_counts()
)

print("\nTesting target distribution:")

print(
    y_test.value_counts()
)


# 10. CHECK DUPLICATE LEAKAGE BETWEEN TRAIN AND TEST
# This is extremely important for our critical analysis.
# We test whether test feature vectors also occur in the training data.

train_rows = set(
    map(
        tuple,
        X_train.to_numpy()
    )
)

test_rows = list(
    map(
        tuple,
        X_test.to_numpy()
    )
)

duplicate_test_matches = sum(
    row in train_rows
    for row in test_rows
)


duplicate_test_percentage = (
    duplicate_test_matches
    /
    len(X_test)
    *
    100
)


print("TRAIN–TEST DUPLICATE LEAKAGE CHECK")

print(
    "Test observations:",
    len(X_test)
)

print(
    "Test observations whose feature vector also appears in training:",
    duplicate_test_matches
)

print(
    "Percentage of test set duplicated in training:",
    round(
        duplicate_test_percentage,
        2
    ),
    "%"
)


# 11. DEFINE PREPROCESSING

scaled_preprocessing = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "scaler",
        StandardScaler()
    )

])


tree_preprocessing = Pipeline([

    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    )

])


# 12. DEFINE PART 1 MODELS

models = {


    # Logistic Regression

    "Logistic Regression":

        Pipeline([

            (
                "preprocessing",
                scaled_preprocessing
            ),

            (
                "classifier",
                LogisticRegression(
                    max_iter=2000,
                    random_state=SEED
                )
            )

        ]),

    # Decision Tree

    "Decision Tree":

        Pipeline([

            (
                "preprocessing",
                tree_preprocessing
            ),

            (
                "classifier",
                DecisionTreeClassifier(
                    random_state=SEED
                )
            )

        ]),


    # Random Forest

    "Random Forest":

        Pipeline([

            (
                "preprocessing",
                tree_preprocessing
            ),

            (
                "classifier",
                RandomForestClassifier(

                    n_estimators=100,

                    random_state=SEED,

                    n_jobs=-1
                )
            )

        ]),


    # XGBoost

    "XGBoost":

        Pipeline([

            (
                "preprocessing",
                tree_preprocessing
            ),

            (
                "classifier",
                XGBClassifier(

                    n_estimators=100,

                    max_depth=6,

                    learning_rate=0.1,

                    objective="binary:logistic",

                    eval_metric="logloss",

                    random_state=SEED,

                    n_jobs=-1
                )
            )

        ]),


    # Naive Bayes

    "Naive Bayes":

        Pipeline([

            (
                "preprocessing",
                scaled_preprocessing
            ),

            (
                "classifier",
                GaussianNB()
            )

        ]),


    # KNN

    "KNN":

        Pipeline([

            (
                "preprocessing",
                scaled_preprocessing
            ),

            (
                "classifier",
                KNeighborsClassifier(
                    n_neighbors=5
                )
            )

        ])
}


# 13. MODEL EVALUATION FUNCTION

def evaluate_model(
    model_name,
    model,
    X_train,
    X_test,
    y_train,
    y_test
):

    model.fit(
        X_train,
        y_train
    )

    predictions = model.predict(
        X_test
    )


    # Probability estimates needed for ROC/AUC

    if hasattr(
        model,
        "predict_proba"
    ):

        probabilities = (
            model
            .predict_proba(
                X_test
            )[:, 1]
        )

    else:

        probabilities = None


    accuracy = accuracy_score(
        y_test,
        predictions
    )


    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )


    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )


    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )


    balanced_accuracy = (
        balanced_accuracy_score(
            y_test,
            predictions
        )
    )


    mcc = matthews_corrcoef(
        y_test,
        predictions
    )


    if probabilities is not None:

        auc_score = roc_auc_score(
            y_test,
            probabilities
        )

    else:

        auc_score = np.nan


    result = {

        "Model":
            model_name,

        "Accuracy":
            accuracy,

        "Precision":
            precision,

        "Recall":
            recall,

        "F1 Score":
            f1,

        "AUC":
            auc_score,

        "Balanced Accuracy":
            balanced_accuracy,

        "MCC":
            mcc,

        "Predictions":
            predictions,

        "Probabilities":
            probabilities,

        "Estimator":
            model
    }


    return result


# 14. TRAIN ALL SIX INDIVIDUAL CLASSIFIERS

print("\n" + "=" * 80)
print("PART 1 — INDIVIDUAL CLASSIFIERS")
print("=" * 80)


individual_results = []

trained_models = {}


for model_name, model in models.items():

    print(
        "\nTraining:",
        model_name
    )

    result = evaluate_model(

        model_name,

        model,

        X_train,
        X_test,

        y_train,
        y_test
    )


    individual_results.append(
        result
    )


    trained_models[
        model_name
    ] = result["Estimator"]


    print(
        f"Accuracy          : "
        f"{result['Accuracy']:.4f}"
    )

    print(
        f"Precision         : "
        f"{result['Precision']:.4f}"
    )

    print(
        f"Recall            : "
        f"{result['Recall']:.4f}"
    )

    print(
        f"F1 Score          : "
        f"{result['F1 Score']:.4f}"
    )

    print(
        f"AUC               : "
        f"{result['AUC']:.4f}"
    )

    print(
        f"Balanced Accuracy : "
        f"{result['Balanced Accuracy']:.4f}"
    )

    print(
        f"MCC               : "
        f"{result['MCC']:.4f}"
    )


# 15. CREATE PART 1 INDIVIDUAL RESULTS TABLE

metric_columns = [

    "Model",

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score",

    "AUC",

    "Balanced Accuracy",

    "MCC"
]


individual_table = pd.DataFrame([

    {

        column:
            result[column]

        for column in metric_columns

    }

    for result in individual_results

])


print("\n" + "=" * 80)
print("PART 1 — INDIVIDUAL MODEL RESULTS")
print("=" * 80)


display(
    individual_table.style.format({

        "Accuracy":
            "{:.4f}",

        "Precision":
            "{:.4f}",

        "Recall":
            "{:.4f}",

        "F1 Score":
            "{:.4f}",

        "AUC":
            "{:.4f}",

        "Balanced Accuracy":
            "{:.4f}",

        "MCC":
            "{:.4f}"
    })
)


individual_table.to_csv(

    f"{OUTPUT_DIR}/part1_individual_results.csv",

    index=False
)


# 16. PART 1 STACKING ENSEMBLE

print("PART 1 — STACKING ENSEMBLE")

stack_estimators = [


    # Logistic Regression

    (
        "lr",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "scaler",
                StandardScaler()
            ),

            (
                "model",
                LogisticRegression(

                    max_iter=2000,

                    random_state=SEED
                )
            )

        ])
    ),


    # Decision Tree

    (
        "dt",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "model",
                DecisionTreeClassifier(
                    random_state=SEED
                )
            )

        ])
    ),


    # Random Forest

    (
        "rf",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "model",
                RandomForestClassifier(

                    n_estimators=100,

                    random_state=SEED,

                    n_jobs=-1
                )
            )

        ])
    ),


    # XGBoost

    (
        "xgb",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "model",
                XGBClassifier(

                    n_estimators=100,

                    max_depth=6,

                    learning_rate=0.1,

                    objective="binary:logistic",

                    eval_metric="logloss",

                    random_state=SEED,

                    n_jobs=-1
                )
            )

        ])
    ),


    # Naive Bayes

    (
        "nb",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "scaler",
                StandardScaler()
            ),

            (
                "model",
                GaussianNB()
            )

        ])
    ),


    # KNN

    (
        "knn",

        Pipeline([

            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),

            (
                "scaler",
                StandardScaler()
            ),

            (
                "model",
                KNeighborsClassifier(
                    n_neighbors=5
                )
            )

        ])
    )
]


stacking_model = StackingClassifier(

    estimators=
        stack_estimators,

    final_estimator=
        LogisticRegression(

            max_iter=2000,

            random_state=SEED
        ),

    stack_method=
        "predict_proba",

    cv=5,

    n_jobs=-1
)


stack_result = evaluate_model(

    "Stacking Ensemble",

    stacking_model,

    X_train,
    X_test,

    y_train,
    y_test
)


print("\nSTACKING RESULTS")

for metric in [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "AUC",
    "Balanced Accuracy",
    "MCC"

]:

    print(
        f"{metric:18s}: "
        f"{stack_result[metric]:.4f}"
    )


# 17. COMPLETE PART 1 RESULTS

part1_results = (
    individual_results
    +
    [stack_result]
)


part1_table = pd.DataFrame([

    {

        column:
            result[column]

        for column in metric_columns

    }

    for result in part1_results

])


print("PART 1 — COMPLETE RESULTS")

display(
    part1_table.style.format({

        "Accuracy":
            "{:.4f}",

        "Precision":
            "{:.4f}",

        "Recall":
            "{:.4f}",

        "F1 Score":
            "{:.4f}",

        "AUC":
            "{:.4f}",

        "Balanced Accuracy":
            "{:.4f}",

        "MCC":
            "{:.4f}"
    })
)


part1_table.to_csv(

    f"{OUTPUT_DIR}/part1_complete_results.csv",

    index=False
)


# 18. PART 1 CONFUSION MATRICES

for result in part1_results:

    cm = confusion_matrix(

        y_test,

        result["Predictions"]
    )


    disp = ConfusionMatrixDisplay(

        confusion_matrix=cm,

        display_labels=[
            "No Disease",
            "Disease"
        ]
    )


    disp.plot(
        values_format="d"
    )


    plt.title(
        f"Confusion Matrix — {result['Model']}"
    )


    filename = (
        result["Model"]
        .lower()
        .replace(
            " ",
            "_"
        )
    )


    plt.tight_layout()


    plt.savefig(

        f"{OUTPUT_DIR}/CM_{filename}.png",

        dpi=300,

        bbox_inches="tight"
    )


    plt.show()


# 19. PART 1 ROC CURVES

plt.figure(
    figsize=(10, 7)
)


for result in part1_results:

    probabilities = (
        result["Probabilities"]
    )


    if probabilities is not None:

        fpr, tpr, thresholds = roc_curve(

            y_test,

            probabilities
        )


        roc_auc = auc(
            fpr,
            tpr
        )


        plt.plot(

            fpr,

            tpr,

            label=(
                f"{result['Model']} "
                f"(AUC={roc_auc:.3f})"
            )
        )


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "ROC Curves — Part 1 Reproduction"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/04_part1_roc_curves.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 20. PART 1 MODEL COMPARISON GRAPH

comparison_plot = (

    part1_table

    .set_index(
        "Model"
    )[

        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score",
            "AUC"
        ]

    ]

)


comparison_plot.plot(

    kind="bar",

    figsize=(13, 7)
)


plt.ylim(
    0,
    1.05
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Model"
)

plt.title(
    "Performance of Reproduced Classification Models"
)

plt.xticks(
    rotation=35,
    ha="right"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/05_part1_model_comparison.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 21. RANDOM FOREST FEATURE IMPORTANCE

rf_pipeline = trained_models[
    "Random Forest"
]


rf_classifier = (

    rf_pipeline

    .named_steps[
        "classifier"
    ]

)


feature_importance = pd.DataFrame({

    "Feature":
        X.columns,

    "Importance":
        rf_classifier.feature_importances_

})


feature_importance = (

    feature_importance

    .sort_values(

        "Importance",

        ascending=False
    )

)


print("RANDOM FOREST FEATURE IMPORTANCE")


display(
    feature_importance
)


feature_importance.to_csv(

    f"{OUTPUT_DIR}/feature_importance.csv",

    index=False
)


plt.figure(
    figsize=(10, 6)
)


plt.barh(

    feature_importance[
        "Feature"
    ][::-1],

    feature_importance[
        "Importance"
    ][::-1]
)


plt.xlabel(
    "Feature Importance"
)

plt.title(
    "Random Forest Feature Importance"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/06_feature_importance.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 22. FIVE-FOLD STRATIFIED CROSS-VALIDATION

print("PART 1 — 5-FOLD STRATIFIED CROSS-VALIDATION")


cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=SEED
)


scoring = {

    "accuracy":
        "accuracy",

    "precision":
        "precision",

    "recall":
        "recall",

    "f1":
        "f1",

    "auc":
        "roc_auc"
}


cv_results = []


models_for_cv = {

    **models,

    "Stacking Ensemble":
        stacking_model

}


for model_name, model in models_for_cv.items():

    print(
        "\nCross-validating:",
        model_name
    )


    scores = cross_validate(

        model,

        X,
        y,

        cv=cv,

        scoring=scoring,

        n_jobs=-1
    )


    cv_results.append({

        "Model":
            model_name,

        "Accuracy Mean":
            scores[
                "test_accuracy"
            ].mean(),

        "Accuracy SD":
            scores[
                "test_accuracy"
            ].std(),

        "Precision Mean":
            scores[
                "test_precision"
            ].mean(),

        "Recall Mean":
            scores[
                "test_recall"
            ].mean(),

        "F1 Mean":
            scores[
                "test_f1"
            ].mean(),

        "AUC Mean":
            scores[
                "test_auc"
            ].mean()
    })


cv_table = pd.DataFrame(
    cv_results
)


print("\nCROSS-VALIDATION RESULTS")


display(
    cv_table.style.format({

        "Accuracy Mean":
            "{:.4f}",

        "Accuracy SD":
            "{:.4f}",

        "Precision Mean":
            "{:.4f}",

        "Recall Mean":
            "{:.4f}",

        "F1 Mean":
            "{:.4f}",

        "AUC Mean":
            "{:.4f}"
    })
)


cv_table.to_csv(

    f"{OUTPUT_DIR}/part1_cross_validation.csv",

    index=False
)


# PART 2

# ROBUST DUPLICATE-AWARE ML EXPERIMENT

# Research motivation:
# The downloaded dataset contains a large number of exact duplicates.
# If identical observations occur in both training and testing data,
# test performance can be artificially optimistic because the model
# may effectively encounter the same feature vector during training.

# PART 2 therefore:

# 1. Removes exact duplicate observations BEFORE splitting.
# 2. Creates a new untouched test set from unique observations.
# 3. Uses stratified CV only on training data.
# 4. Tunes RF and XGBoost hyperparameters.
# 5. Creates a tuned stacking ensemble.
# 6. Evaluates once on the unseen unique test set.


print("PART 2 — DUPLICATE-AWARE ROBUST ML PIPELINE")


# 23. REMOVE EXACT DUPLICATES

original_n = len(
    df
)


df_unique = (

    df

    .drop_duplicates()

    .reset_index(
        drop=True
    )

)


unique_n = len(
    df_unique
)


removed_n = (
    original_n
    -
    unique_n
)


removed_percentage = (

    removed_n

    /
    original_n

    *
    100
)


print(
    "\nOriginal observations:",
    original_n
)

print(
    "Unique observations:",
    unique_n
)

print(
    "Exact duplicates removed:",
    removed_n
)

print(
    "Percentage removed:",
    round(
        removed_percentage,
        2
    ),
    "%"
)


print(
    "\nUnique-data target distribution:"
)

print(
    df_unique[
        "target"
    ].value_counts()
)


print(
    "\nUnique-data target proportions:"
)

print(
    df_unique[
        "target"
    ]
    .value_counts(
        normalize=True
    )
    .round(4)
)


# 24. UNIQUE X AND y

X_unique = (

    df_unique

    .drop(
        columns=[
            "target"
        ]
    )

)


y_unique = (

    df_unique[
        "target"
    ]

    .astype(int)

)


# 25. DUPLICATE-AWARE TRAIN TEST SPLIT

X_train_u, X_test_u, y_train_u, y_test_u = train_test_split(

    X_unique,

    y_unique,

    test_size=0.20,

    random_state=SEED,

    stratify=y_unique
)


print("PART 2 — UNIQUE DATA SPLIT")

print(
    "Training observations:",
    len(X_train_u)
)

print(
    "Testing observations:",
    len(X_test_u)
)

# 26. VERIFY ZERO EXACT TRAIN–TEST OVERLAP

train_unique_set = set(

    map(
        tuple,

        X_train_u.to_numpy()
    )

)


test_unique_rows = list(

    map(
        tuple,

        X_test_u.to_numpy()
    )

)


unique_overlap = sum(

    row in train_unique_set

    for row in test_unique_rows
)


print(
    "\nExact feature-vector overlap between "
    "Part 2 training and testing sets:",
    unique_overlap
)


if unique_overlap == 0:

    print(
        "✓ No exact train-test duplicate leakage."
    )

else:

    print(
        "WARNING: overlap detected."
    )


# 27. BASELINE PERFORMANCE ON UNIQUE DATA

# Before tuning anything, evaluate the Part 1 stacking approach on
# the deduplicated dataset.
#
# This isolates the effect of duplicate removal.

print("PART 2A — ORIGINAL STACKING ON UNIQUE DATA")

unique_baseline_result = evaluate_model(

    "Original Stacking — Unique Data",

    stacking_model,

    X_train_u,
    X_test_u,

    y_train_u,
    y_test_u
)


for metric in [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "AUC",
    "Balanced Accuracy",
    "MCC"

]:

    print(
        f"{metric:18s}: "
        f"{unique_baseline_result[metric]:.4f}"
    )


# 28. CROSS-VALIDATION OBJECT FOR TRAINING DATA ONLY

inner_cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=SEED
)


# 29. RANDOM FOREST HYPERPARAMETER OPTIMISATION

print("PART 2B — TUNING RANDOM FOREST")


rf_tuning_pipeline = Pipeline([

    (
        "imputer",

        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "model",

        RandomForestClassifier(

            random_state=SEED,

            n_jobs=-1
        )
    )

])


rf_parameter_space = {

    "model__n_estimators":
        [
            100,
            200,
            300,
            500
        ],

    "model__max_depth":
        [
            None,
            3,
            5,
            7,
            10,
            15
        ],

    "model__min_samples_split":
        [
            2,
            4,
            6,
            10
        ],

    "model__min_samples_leaf":
        [
            1,
            2,
            4,
            6
        ],

    "model__max_features":
        [
            "sqrt",
            "log2",
            None
        ]

}


rf_search = RandomizedSearchCV(

    estimator=
        rf_tuning_pipeline,

    param_distributions=
        rf_parameter_space,

    n_iter=30,

    scoring=
        "roc_auc",

    cv=
        inner_cv,

    random_state=
        SEED,

    n_jobs=
        -1,

    verbose=
        0,

    refit=True
)


rf_search.fit(

    X_train_u,

    y_train_u
)


print(
    "\nBest RF training CV AUC:",
    round(
        rf_search.best_score_,
        4
    )
)


print(
    "\nBest Random Forest parameters:"
)

print(
    rf_search.best_params_
)


best_rf = (
    rf_search.best_estimator_
)


# 30. XGBOOST HYPERPARAMETER OPTIMISATION

print("PART 2C — TUNING XGBOOST")

xgb_tuning_pipeline = Pipeline([

    (
        "imputer",

        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "model",

        XGBClassifier(

            objective=
                "binary:logistic",

            eval_metric=
                "logloss",

            random_state=
                SEED,

            n_jobs=
                -1
        )
    )

])


xgb_parameter_space = {

    "model__n_estimators":
        [
            50,
            100,
            150,
            200,
            300
        ],

    "model__max_depth":
        [
            2,
            3,
            4,
            5,
            6
        ],

    "model__learning_rate":
        [
            0.01,
            0.03,
            0.05,
            0.1,
            0.2
        ],

    "model__subsample":
        [
            0.7,
            0.8,
            0.9,
            1.0
        ],

    "model__colsample_bytree":
        [
            0.7,
            0.8,
            0.9,
            1.0
        ],

    "model__min_child_weight":
        [
            1,
            2,
            3,
            5
        ]

}


xgb_search = RandomizedSearchCV(

    estimator=
        xgb_tuning_pipeline,

    param_distributions=
        xgb_parameter_space,

    n_iter=
        30,

    scoring=
        "roc_auc",

    cv=
        inner_cv,

    random_state=
        SEED,

    n_jobs=
        -1,

    verbose=
        0,

    refit=
        True
)


xgb_search.fit(

    X_train_u,

    y_train_u
)


print(
    "\nBest XGBoost training CV AUC:",
    round(
        xgb_search.best_score_,
        4
    )
)


print(
    "\nBest XGBoost parameters:"
)

print(
    xgb_search.best_params_
)


best_xgb = (
    xgb_search.best_estimator_
)


# 31. LOGISTIC REGRESSION COMPONENT

robust_lr = Pipeline([

    (
        "imputer",

        SimpleImputer(
            strategy="median"
        )
    ),

    (
        "scaler",

        StandardScaler()
    ),

    (
        "model",

        LogisticRegression(

            C=1.0,

            max_iter=3000,

            random_state=SEED
        )
    )

])


# 32. PROPOSED TUNED STACKING ENSEMBLE

# We use three complementary models:

# Logistic Regression: linear decision boundary

# Random Forest: bagged nonlinear trees

# XGBoost: sequential gradient-boosted trees

# This gives the stack heterogeneous model behaviour while avoiding
# unnecessary weak/redundant learners.


proposed_estimators = [

    (
        "rf",
        best_rf
    ),

    (
        "xgb",
        best_xgb
    ),

    (
        "lr",
        robust_lr
    )

]


proposed_model = StackingClassifier(

    estimators=
        proposed_estimators,

    final_estimator=
        LogisticRegression(

            C=0.5,

            max_iter=3000,

            random_state=SEED
        ),

    stack_method=
        "predict_proba",

    cv=
        inner_cv,

    passthrough=
        True,

    n_jobs=
        -1
)


# 33. FINAL PART 2 TEST

print("PART 2D — FINAL PROPOSED MODEL")



proposed_result = evaluate_model(

    "Proposed Duplicate-Aware Tuned Stacking",

    proposed_model,

    X_train_u,
    X_test_u,

    y_train_u,
    y_test_u
)


for metric in [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "AUC",
    "Balanced Accuracy",
    "MCC"

]:

    print(
        f"{metric:18s}: "
        f"{proposed_result[metric]:.4f}"
    )


# 34. PROPOSED MODEL CLASSIFICATION REPORT

print("PART 2 CLASSIFICATION REPORT")

print(

    classification_report(

        y_test_u,

        proposed_result[
            "Predictions"
        ],

        target_names=[

            "No Heart Disease",

            "Heart Disease"

        ],

        digits=4
    )

)


report_dictionary = classification_report(

    y_test_u,

    proposed_result[
        "Predictions"
    ],

    output_dict=True
)


pd.DataFrame(

    report_dictionary

).T.to_csv(

    f"{OUTPUT_DIR}/part2_classification_report.csv"
)


# 35. PART 2 CONFUSION MATRIX

part2_cm = confusion_matrix(

    y_test_u,

    proposed_result[
        "Predictions"
    ]

)


disp = ConfusionMatrixDisplay(

    confusion_matrix=
        part2_cm,

    display_labels=[
        "No Disease",
        "Disease"
    ]
)


disp.plot(
    values_format="d"
)


plt.title(
    "Part 2 — Proposed Duplicate-Aware Model"
)


plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/07_part2_confusion_matrix.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 36. THREE-WAY COMPARISON

# A = Part 1 model on duplicated dataset

# B = Same general stacking approach on deduplicated dataset

# C = Proposed tuned model on deduplicated dataset

# This comparison allows us to separate:

# Effect of duplicate removal
# from
#
# Effect of our modelling improvements.

three_way_comparison = pd.DataFrame([


    {

        "Experiment":
            "Part 1 Stacking — Original 1025 Rows",

        "Accuracy":
            stack_result[
                "Accuracy"
            ],

        "Precision":
            stack_result[
                "Precision"
            ],

        "Recall":
            stack_result[
                "Recall"
            ],

        "F1 Score":
            stack_result[
                "F1 Score"
            ],

        "AUC":
            stack_result[
                "AUC"
            ],

        "Balanced Accuracy":
            stack_result[
                "Balanced Accuracy"
            ],

        "MCC":
            stack_result[
                "MCC"
            ]

    },


    {

        "Experiment":
            "Original Stacking — Unique Rows",

        "Accuracy":
            unique_baseline_result[
                "Accuracy"
            ],

        "Precision":
            unique_baseline_result[
                "Precision"
            ],

        "Recall":
            unique_baseline_result[
                "Recall"
            ],

        "F1 Score":
            unique_baseline_result[
                "F1 Score"
            ],

        "AUC":
            unique_baseline_result[
                "AUC"
            ],

        "Balanced Accuracy":
            unique_baseline_result[
                "Balanced Accuracy"
            ],

        "MCC":
            unique_baseline_result[
                "MCC"
            ]

    },


    {

        "Experiment":
            "Proposed Tuned Stacking — Unique Rows",

        "Accuracy":
            proposed_result[
                "Accuracy"
            ],

        "Precision":
            proposed_result[
                "Precision"
            ],

        "Recall":
            proposed_result[
                "Recall"
            ],

        "F1 Score":
            proposed_result[
                "F1 Score"
            ],

        "AUC":
            proposed_result[
                "AUC"
            ],

        "Balanced Accuracy":
            proposed_result[
                "Balanced Accuracy"
            ],

        "MCC":
            proposed_result[
                "MCC"
            ]

    }

])


print("FINAL THREE-WAY EXPERIMENTAL COMPARISON")


display(

    three_way_comparison.style.format({

        "Accuracy":
            "{:.4f}",

        "Precision":
            "{:.4f}",

        "Recall":
            "{:.4f}",

        "F1 Score":
            "{:.4f}",

        "AUC":
            "{:.4f}",

        "Balanced Accuracy":
            "{:.4f}",

        "MCC":
            "{:.4f}"

    })

)


three_way_comparison.to_csv(

    f"{OUTPUT_DIR}/FINAL_THREE_WAY_COMPARISON.csv",

    index=False
)


# 37. THREE-WAY COMPARISON GRAPH

plot_comparison = (

    three_way_comparison

    .set_index(
        "Experiment"
    )[

        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score",
            "AUC"
        ]

    ]

)


plot_comparison.plot(

    kind="bar",

    figsize=(13, 7)
)


plt.ylim(
    0,
    1.05
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Experimental Setting"
)

plt.title(
    "Effect of Duplicate Removal and Model Improvement"
)

plt.xticks(
    rotation=20,
    ha="right"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/08_three_way_comparison.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 38. ROC — UNIQUE BASELINE vs PROPOSED MODEL

plt.figure(
    figsize=(8, 6)
)


# Unique baseline

fpr_baseline, tpr_baseline, _ = roc_curve(

    y_test_u,

    unique_baseline_result[
        "Probabilities"
    ]

)


auc_baseline = roc_auc_score(

    y_test_u,

    unique_baseline_result[
        "Probabilities"
    ]

)


plt.plot(

    fpr_baseline,

    tpr_baseline,

    label=(
        "Original Stacking on Unique Data "
        f"(AUC={auc_baseline:.3f})"
    )
)


# Proposed model

fpr_proposed, tpr_proposed, _ = roc_curve(

    y_test_u,

    proposed_result[
        "Probabilities"
    ]

)


auc_proposed = roc_auc_score(

    y_test_u,

    proposed_result[
        "Probabilities"
    ]

)


plt.plot(

    fpr_proposed,

    tpr_proposed,

    label=(
        "Proposed Tuned Stacking "
        f"(AUC={auc_proposed:.3f})"
    )
)


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"
)


plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Part 2 ROC Comparison on Deduplicated Data"
)

plt.legend(
    loc="lower right"
)

plt.tight_layout()


plt.savefig(

    f"{OUTPUT_DIR}/09_part2_roc_comparison.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# 39. QUANTIFY DUPLICATE EFFECT

duplicate_effect = pd.DataFrame({

    "Metric": [

        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "AUC",
        "Balanced Accuracy",
        "MCC"

    ],

    "Original 1025-Row Stacking": [

        stack_result[
            "Accuracy"
        ],

        stack_result[
            "Precision"
        ],

        stack_result[
            "Recall"
        ],

        stack_result[
            "F1 Score"
        ],

        stack_result[
            "AUC"
        ],

        stack_result[
            "Balanced Accuracy"
        ],

        stack_result[
            "MCC"
        ]

    ],

    "Deduplicated Stacking": [

        unique_baseline_result[
            "Accuracy"
        ],

        unique_baseline_result[
            "Precision"
        ],

        unique_baseline_result[
            "Recall"
        ],

        unique_baseline_result[
            "F1 Score"
        ],

        unique_baseline_result[
            "AUC"
        ],

        unique_baseline_result[
            "Balanced Accuracy"
        ],

        unique_baseline_result[
            "MCC"
        ]

    ]

})


duplicate_effect[
    "Change After Deduplication"
] = (

    duplicate_effect[
        "Deduplicated Stacking"
    ]

    -

    duplicate_effect[
        "Original 1025-Row Stacking"
    ]

)


print("EFFECT OF DUPLICATE REMOVAL")

display(

    duplicate_effect.style.format({

        "Original 1025-Row Stacking":
            "{:.4f}",

        "Deduplicated Stacking":
            "{:.4f}",

        "Change After Deduplication":
            "{:+.4f}"

    })

)


duplicate_effect.to_csv(

    f"{OUTPUT_DIR}/duplicate_effect.csv",

    index=False
)


# ======================================================================
# 40. QUANTIFY IMPROVEMENT ON FAIR UNIQUE-DATA COMPARISON
# ======================================================================

fair_improvement = pd.DataFrame({

    "Metric": [

        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "AUC",
        "Balanced Accuracy",
        "MCC"

    ],

    "Original Stacking": [

        unique_baseline_result[
            "Accuracy"
        ],

        unique_baseline_result[
            "Precision"
        ],

        unique_baseline_result[
            "Recall"
        ],

        unique_baseline_result[
            "F1 Score"
        ],

        unique_baseline_result[
            "AUC"
        ],

        unique_baseline_result[
            "Balanced Accuracy"
        ],

        unique_baseline_result[
            "MCC"
        ]

    ],

    "Proposed Tuned Stacking": [

        proposed_result[
            "Accuracy"
        ],

        proposed_result[
            "Precision"
        ],

        proposed_result[
            "Recall"
        ],

        proposed_result[
            "F1 Score"
        ],

        proposed_result[
            "AUC"
        ],

        proposed_result[
            "Balanced Accuracy"
        ],

        proposed_result[
            "MCC"
        ]

    ]

})


fair_improvement[
    "Improvement"
] = (

    fair_improvement[
        "Proposed Tuned Stacking"
    ]

    -

    fair_improvement[
        "Original Stacking"
    ]

)


print("FAIR PART 2 MODEL IMPROVEMENT")


display(

    fair_improvement.style.format({

        "Original Stacking":
            "{:.4f}",

        "Proposed Tuned Stacking":
            "{:.4f}",

        "Improvement":
            "{:+.4f}"

    })

)


fair_improvement.to_csv(

    f"{OUTPUT_DIR}/part2_fair_improvement.csv",

    index=False
)


# 41. SAVE IMPORTANT EXPERIMENT INFORMATION

experiment_information = {

    "random_seed":
        SEED,

    "original_dataset_rows":
        int(
            len(df)
        ),

    "original_predictor_count":
        int(
            X.shape[1]
        ),

    "positive_class_original":
        int(
            y.sum()
        ),

    "negative_class_original":
        int(
            len(y) - y.sum()
        ),

    "exact_duplicates":
        int(
            df.duplicated().sum()
        ),

    "unique_rows":
        int(
            len(df_unique)
        ),

    "duplicate_percentage":
        float(
            removed_percentage
        ),

    "part1_train_rows":
        int(
            len(X_train)
        ),

    "part1_test_rows":
        int(
            len(X_test)
        ),

    "part1_test_rows_also_seen_in_training":
        int(
            duplicate_test_matches
        ),

    "part1_test_overlap_percentage":
        float(
            duplicate_test_percentage
        ),

    "part2_train_rows":
        int(
            len(X_train_u)
        ),

    "part2_test_rows":
        int(
            len(X_test_u)
        ),

    "part2_train_test_exact_overlap":
        int(
            unique_overlap
        ),

    "part1_test_size":
        0.20,

    "part2_test_size":
        0.20,

    "cross_validation":
        "5-fold stratified",

    "rf_best_cv_auc":
        float(
            rf_search.best_score_
        ),

    "rf_best_parameters":
        rf_search.best_params_,

    "xgb_best_cv_auc":
        float(
            xgb_search.best_score_
        ),

    "xgb_best_parameters":
        xgb_search.best_params_

}


with open(

    f"{OUTPUT_DIR}/experiment_information.json",

    "w"

) as file:

    json.dump(

        experiment_information,

        file,

        indent=4

    )


# 42. MASTER RESULTS

master_results = pd.concat(

    [

        part1_table,

        pd.DataFrame([

            {

                "Model":
                    "Original Stacking — Unique Data",

                "Accuracy":
                    unique_baseline_result[
                        "Accuracy"
                    ],

                "Precision":
                    unique_baseline_result[
                        "Precision"
                    ],

                "Recall":
                    unique_baseline_result[
                        "Recall"
                    ],

                "F1 Score":
                    unique_baseline_result[
                        "F1 Score"
                    ],

                "AUC":
                    unique_baseline_result[
                        "AUC"
                    ],

                "Balanced Accuracy":
                    unique_baseline_result[
                        "Balanced Accuracy"
                    ],

                "MCC":
                    unique_baseline_result[
                        "MCC"
                    ]

            },

            {

                "Model":
                    "Proposed Duplicate-Aware Tuned Stacking",

                "Accuracy":
                    proposed_result[
                        "Accuracy"
                    ],

                "Precision":
                    proposed_result[
                        "Precision"
                    ],

                "Recall":
                    proposed_result[
                        "Recall"
                    ],

                "F1 Score":
                    proposed_result[
                        "F1 Score"
                    ],

                "AUC":
                    proposed_result[
                        "AUC"
                    ],

                "Balanced Accuracy":
                    proposed_result[
                        "Balanced Accuracy"
                    ],

                "MCC":
                    proposed_result[
                        "MCC"
                    ]

            }

        ])

    ],

    ignore_index=True
)


master_results.to_csv(

    f"{OUTPUT_DIR}/MASTER_RESULTS.csv",

    index=False
)


print("MASTER RESULTS TABLE")


display(

    master_results.style.format({

        "Accuracy":
            "{:.4f}",

        "Precision":
            "{:.4f}",

        "Recall":
            "{:.4f}",

        "F1 Score":
            "{:.4f}",

        "AUC":
            "{:.4f}",

        "Balanced Accuracy":
            "{:.4f}",

        "MCC":
            "{:.4f}"

    })

)


# 43. FINAL SUMMARY

print("FINAL EXPERIMENT SUMMARY")


print("\nDATASET")

print(
    "Original observations:",
    len(df)
)

print(
    "Exact duplicates:",
    df.duplicated().sum()
)

print(
    "Unique observations:",
    len(df_unique)
)

print(
    "Duplicate percentage:",
    f"{removed_percentage:.2f}%"
)


print("\nPART 1 DATA LEAKAGE CHECK")

print(
    "Test observations also represented in training:",
    duplicate_test_matches,
    "/",
    len(X_test)
)

print(
    "Overlap percentage:",
    f"{duplicate_test_percentage:.2f}%"
)


print("\nPART 1 — REPRODUCED STACKING")

print(
    f"Accuracy : "
    f"{stack_result['Accuracy']:.4f}"
)

print(
    f"Precision: "
    f"{stack_result['Precision']:.4f}"
)

print(
    f"Recall   : "
    f"{stack_result['Recall']:.4f}"
)

print(
    f"F1 Score : "
    f"{stack_result['F1 Score']:.4f}"
)

print(
    f"AUC      : "
    f"{stack_result['AUC']:.4f}"
)


print("\nPART 2A — STACKING AFTER DEDUPLICATION")

print(
    f"Accuracy : "
    f"{unique_baseline_result['Accuracy']:.4f}"
)

print(
    f"Precision: "
    f"{unique_baseline_result['Precision']:.4f}"
)

print(
    f"Recall   : "
    f"{unique_baseline_result['Recall']:.4f}"
)

print(
    f"F1 Score : "
    f"{unique_baseline_result['F1 Score']:.4f}"
)

print(
    f"AUC      : "
    f"{unique_baseline_result['AUC']:.4f}"
)


print("\nPART 2B — PROPOSED TUNED STACKING")

print(
    f"Accuracy : "
    f"{proposed_result['Accuracy']:.4f}"
)

print(
    f"Precision: "
    f"{proposed_result['Precision']:.4f}"
)

print(
    f"Recall   : "
    f"{proposed_result['Recall']:.4f}"
)

print(
    f"F1 Score : "
    f"{proposed_result['F1 Score']:.4f}"
)

print(
    f"AUC      : "
    f"{proposed_result['AUC']:.4f}"
)


print("\nBEST RANDOM FOREST PARAMETERS")

print(
    rf_search.best_params_
)


print("\nBEST XGBOOST PARAMETERS")

print(
    xgb_search.best_params_
)


print("FILES CREATED FOR RESEARCH REPORT")


for filename in sorted(
    os.listdir(
        OUTPUT_DIR
    )
):

    print(
        "•",
        filename
    )


print("ALL EXPERIMENTS COMPLETED SUCCESSFULLY")
```

```python
# FINAL ADDITIONAL EXPERIMENT
# REPEATED STRATIFIED CROSS-VALIDATION ON DEDUPLICATED DATA

# 5 folds x 3 repeats = 15 paired evaluations per model

# Run AFTER your previous notebook code.

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    balanced_accuracy_score,
    matthews_corrcoef
)
from scipy.stats import wilcoxon

OUTPUT_DIR = "/content/HD_Heart_Attack_Research"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# 1. CHECK THAT PREVIOUS EXPERIMENTS ARE AVAILABLE

required_variables = [
    "df_unique",
    "stacking_model",
    "proposed_model",
    "part1_table",
    "cv_table"
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise RuntimeError(
        "Your previous Colab variables are missing: "
        + ", ".join(missing_variables)
        + ". Run the earlier experiment cells first."
    )

X_cv = df_unique.drop(columns=["target"])
y_cv = df_unique["target"].astype(int)

assert df_unique.duplicated().sum() == 0, (
    "Duplicate rows remain in the validation dataset."
)

print("REPEATED VALIDATION ON UNIQUE HEART DISEASE OBSERVATIONS")

print("Unique observations:", len(df_unique))
print("Predictors:", X_cv.shape[1])
print("Exact duplicates:", df_unique.duplicated().sum())

# 2. DEFINE REPEATED STRATIFIED CV

N_SPLITS = 5
N_REPEATS = 3
SEED = 42

repeated_cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=SEED
)

# Reuse the previously selected hyperparameters.
# No new hyperparameter search is performed here.

comparison_models = {
    "Original Stacking": stacking_model,
    "Proposed Tuned Stacking": proposed_model
}

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "AUC",
    "Balanced Accuracy",
    "MCC"
]

fold_results = []

# 3. EVALUATE BOTH MODELS ON IDENTICAL FOLDS

splits = list(repeated_cv.split(X_cv, y_cv))

for fold_number, (train_idx, test_idx) in enumerate(splits, start=1):

    X_fold_train = X_cv.iloc[train_idx]
    X_fold_test = X_cv.iloc[test_idx]

    y_fold_train = y_cv.iloc[train_idx]
    y_fold_test = y_cv.iloc[test_idx]

    print(
        f"\nFold {fold_number:02d}/{len(splits)} "
        f"| Train: {len(train_idx)} "
        f"| Test: {len(test_idx)}"
    )

    # Confirm that identical feature vectors do not cross the fold boundary.
    train_vectors = set(
        map(tuple, X_fold_train.to_numpy())
    )

    test_vectors = set(
        map(tuple, X_fold_test.to_numpy())
    )

    overlap = len(train_vectors.intersection(test_vectors))

    if overlap != 0:
        raise RuntimeError(
            f"Duplicate feature-vector overlap detected in fold "
            f"{fold_number}: {overlap}"
        )

    for model_name, model_template in comparison_models.items():

        # A fresh model is trained for each fold.
        model = clone(model_template)

        model.fit(X_fold_train, y_fold_train)

        predictions = model.predict(X_fold_test)

        probabilities = model.predict_proba(
            X_fold_test
        )[:, 1]

        row = {
            "Fold": fold_number,
            "Repeat": (fold_number - 1) // N_SPLITS + 1,
            "Model": model_name,

            "Accuracy": accuracy_score(
                y_fold_test, predictions
            ),

            "Precision": precision_score(
                y_fold_test,
                predictions,
                zero_division=0
            ),

            "Recall": recall_score(
                y_fold_test,
                predictions,
                zero_division=0
            ),

            "F1": f1_score(
                y_fold_test,
                predictions,
                zero_division=0
            ),

            "AUC": roc_auc_score(
                y_fold_test,
                probabilities
            ),

            "Balanced Accuracy": balanced_accuracy_score(
                y_fold_test,
                predictions
            ),

            "MCC": matthews_corrcoef(
                y_fold_test,
                predictions
            )
        }

        fold_results.append(row)

        print(
            f"  {model_name:25s} "
            f"Accuracy={row['Accuracy']:.4f} "
            f"F1={row['F1']:.4f} "
            f"AUC={row['AUC']:.4f}"
        )

# 4. SAVE ALL 30 MODEL-FOLD RESULTS

fold_df = pd.DataFrame(fold_results)

fold_df.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_all_folds.csv",
    index=False
)

# 5. MEAN AND STANDARD DEVIATION

summary_rows = []

for model_name in comparison_models:

    model_rows = fold_df[
        fold_df["Model"] == model_name
    ]

    row = {"Model": model_name}

    for metric in metrics:

        row[f"{metric} Mean"] = model_rows[metric].mean()

        row[f"{metric} SD"] = model_rows[metric].std(ddof=1)

        row[f"{metric} Mean ± SD"] = (
            f"{model_rows[metric].mean():.4f} ± "
            f"{model_rows[metric].std(ddof=1):.4f}"
        )

    summary_rows.append(row)

summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_summary.csv",
    index=False
)

print("REPEATED CROSS-VALIDATION: MEAN ± STANDARD DEVIATION")

display(
    summary_df[
        ["Model"] + [
            f"{metric} Mean ± SD"
            for metric in metrics
        ]
    ]
)

# 6. PAIRED FOLD DIFFERENCES

paired_rows = []

for metric in metrics:

    paired = fold_df.pivot(
        index="Fold",
        columns="Model",
        values=metric
    )

    baseline = paired["Original Stacking"].to_numpy()

    proposed = paired["Proposed Tuned Stacking"].to_numpy()

    differences = proposed - baseline

    paired_rows.append({
        "Metric": metric,
        "Baseline Mean": baseline.mean(),
        "Proposed Mean": proposed.mean(),
        "Mean Paired Difference": differences.mean(),
        "Difference SD": differences.std(ddof=1),
        "Proposed Better Folds": int((differences > 0).sum()),
        "Equal Folds": int((differences == 0).sum()),
        "Baseline Better Folds": int((differences < 0).sum())
    })

paired_df = pd.DataFrame(paired_rows)

paired_df.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_paired_differences.csv",
    index=False
)

print("PAIRED COMPARISON ON IDENTICAL VALIDATION FOLDS")

display(paired_df.round(4))

# 7. EXPLORATORY WILCOXON TEST
#
# Repeated CV folds are not statistically independent.
# These p-values are exploratory only, NOT confirmatory evidence.
# We will not claim statistical significance from them.

test_rows = []

for metric in ["Accuracy", "F1", "AUC"]:

    paired = fold_df.pivot(
        index="Fold",
        columns="Model",
        values=metric
    )

    baseline = paired["Original Stacking"].to_numpy()

    proposed = paired["Proposed Tuned Stacking"].to_numpy()

    differences = proposed - baseline

    if np.allclose(differences, 0):
        p_value = 1.0
    else:
        p_value = wilcoxon(
            proposed,
            baseline,
            zero_method="wilcox",
            alternative="two-sided"
        ).pvalue

    test_rows.append({
        "Metric": metric,
        "Exploratory Wilcoxon p-value": p_value
    })

test_df = pd.DataFrame(test_rows)

test_df.to_csv(
    f"{OUTPUT_DIR}/exploratory_wilcoxon.csv",
    index=False
)

print("\nEXPLORATORY PAIRED TESTS")
display(test_df.round(5))

print(
    "\nInterpretation warning: repeated cross-validation folds "
    "are dependent. Do not interpret these p-values as "
    "confirmatory statistical significance."
)

# 8. PLOT ACCURACY, F1 AND AUC ACROSS FOLDS

for metric in ["Accuracy", "F1", "AUC"]:

    plot_df = fold_df.pivot(
        index="Fold",
        columns="Model",
        values=metric
    )

    ax = plot_df.plot(
        marker="o",
        figsize=(11, 5)
    )

    ax.set_xlabel("Validation Fold")
    ax.set_ylabel(metric)
    ax.set_title(
        f"Repeated Stratified CV — {metric}"
    )

    ax.set_ylim(0, 1.05)

    plt.xticks(
        range(1, len(splits) + 1)
    )

    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()

    plt.savefig(
        f"{OUTPUT_DIR}/repeated_cv_{metric.lower().replace(' ', '_')}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

# 9. PRINT PREVIOUS RESULTS AS PLAIN TEXT

print("PART 1 — ORIGINAL HOLD-OUT RESULTS")

print(
    part1_table.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("PART 1 — PREVIOUS 5-FOLD CV RESULTS")
```

```python
# FINAL ROBUSTNESS + STATISTICAL VALIDATION


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from scipy.stats import wilcoxon

print("FINAL ROBUSTNESS AND STATISTICAL VALIDATION")

# 1. USE ONLY DEDUPLICATED DATA

X_robust = df_unique.drop(columns=["target"])
y_robust = df_unique["target"].astype(int)

print("\nDataset used for robust validation:")
print("Observations:", len(X_robust))
print("Predictors:", X_robust.shape[1])
print("Exact duplicates:", df_unique.duplicated().sum())


# 2. REPEATED STRATIFIED CROSS-VALIDATION

# 5 folds × 10 repeats = 50 evaluations.
# Both models receive EXACTLY the same folds.

repeated_cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=42
)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "auc": "roc_auc"
}


# 3. ORIGINAL STACKING ON DEDUPLICATED DATA

print("\nRunning repeated CV for ORIGINAL stacking model...")

original_scores = cross_validate(
    stacking_model,
    X_robust,
    y_robust,
    cv=repeated_cv,
    scoring=scoring,
    n_jobs=-1
)

print("✓ Original stacking complete")


# 4. PROPOSED TUNED STACKING ON DEDUPLICATED DATA

print("\nRunning repeated CV for PROPOSED tuned stacking model...")

proposed_scores = cross_validate(
    proposed_model,
    X_robust,
    y_robust,
    cv=repeated_cv,
    scoring=scoring,
    n_jobs=-1
)

print("✓ Proposed stacking complete")


# 5. SUMMARY FUNCTION

def summarise_cv(scores, model_name):

    return {
        "Model": model_name,

        "Accuracy Mean":
            np.mean(scores["test_accuracy"]),

        "Accuracy SD":
            np.std(scores["test_accuracy"]),

        "Precision Mean":
            np.mean(scores["test_precision"]),

        "Precision SD":
            np.std(scores["test_precision"]),

        "Recall Mean":
            np.mean(scores["test_recall"]),

        "Recall SD":
            np.std(scores["test_recall"]),

        "F1 Mean":
            np.mean(scores["test_f1"]),

        "F1 SD":
            np.std(scores["test_f1"]),

        "AUC Mean":
            np.mean(scores["test_auc"]),

        "AUC SD":
            np.std(scores["test_auc"])
    }


original_summary = summarise_cv(
    original_scores,
    "Original Stacking"
)

proposed_summary = summarise_cv(
    proposed_scores,
    "Proposed Tuned Stacking"
)


robust_table = pd.DataFrame([
    original_summary,
    proposed_summary
])


print("REPEATED 5-FOLD CV RESULTS — 10 REPEATS / 50 TEST FOLDS")

print(
    robust_table.round(4).to_string(index=False)
)


# 6. PRINT REPORT-FRIENDLY MEAN ± SD

print("REPORT-FRIENDLY RESULTS")

for model_name, scores in [
    ("Original Stacking", original_scores),
    ("Proposed Tuned Stacking", proposed_scores)
]:

    print(f"\n{model_name}")

    print(
        "Accuracy : "
        f"{np.mean(scores['test_accuracy']):.4f} "
        f"± {np.std(scores['test_accuracy']):.4f}"
    )

    print(
        "Precision: "
        f"{np.mean(scores['test_precision']):.4f} "
        f"± {np.std(scores['test_precision']):.4f}"
    )

    print(
        "Recall   : "
        f"{np.mean(scores['test_recall']):.4f} "
        f"± {np.std(scores['test_recall']):.4f}"
    )

    print(
        "F1 Score : "
        f"{np.mean(scores['test_f1']):.4f} "
        f"± {np.std(scores['test_f1']):.4f}"
    )

    print(
        "AUC      : "
        f"{np.mean(scores['test_auc']):.4f} "
        f"± {np.std(scores['test_auc']):.4f}"
    )



# 7. ABSOLUTE MEAN DIFFERENCES

metrics = {
    "Accuracy": "test_accuracy",
    "Precision": "test_precision",
    "Recall": "test_recall",
    "F1 Score": "test_f1",
    "AUC": "test_auc"
}

improvement_rows = []

for metric_name, key in metrics.items():

    original_mean = np.mean(
        original_scores[key]
    )

    proposed_mean = np.mean(
        proposed_scores[key]
    )

    difference = (
        proposed_mean
        -
        original_mean
    )

    improvement_rows.append({
        "Metric": metric_name,
        "Original Mean": original_mean,
        "Proposed Mean": proposed_mean,
        "Absolute Difference": difference
    })


improvement_table = pd.DataFrame(
    improvement_rows
)

print("MEAN PERFORMANCE DIFFERENCE")

print(
    improvement_table
    .round(4)
    .to_string(index=False)
)


# 8. PAIRED WILCOXON SIGNED-RANK TEST

# H0:
# No systematic difference in fold-wise performance.

# H1:
# Fold-wise performance differs between the two approaches.

# We use the SAME 50 folds for both models.

print("PAIRED WILCOXON SIGNED-RANK TEST")

statistical_results = []

for metric_name, key in metrics.items():

    original_values = original_scores[key]
    proposed_values = proposed_scores[key]

    differences = (
        proposed_values
        -
        original_values
    )

    # Wilcoxon can fail if every paired difference is exactly zero.
    if np.allclose(differences, 0):

        statistic = 0.0
        p_value = 1.0

    else:

        statistic, p_value = wilcoxon(
            proposed_values,
            original_values,
            alternative="two-sided",
            zero_method="wilcox"
        )

    statistical_results.append({

        "Metric": metric_name,

        "Wilcoxon Statistic":
            statistic,

        "p-value":
            p_value,

        "Significant at 0.05":
            "Yes" if p_value < 0.05 else "No"
    })


stats_table = pd.DataFrame(
    statistical_results
)


print(
    stats_table
    .round(6)
    .to_string(index=False)
)

# 9. 95% CONFIDENCE INTERVALS

# Descriptive normal-approximation CI across the 50 fold scores.

print("95% CONFIDENCE INTERVALS")


ci_rows = []

for model_name, scores in [

    ("Original Stacking", original_scores),

    ("Proposed Tuned Stacking", proposed_scores)

]:

    for metric_name, key in metrics.items():

        values = scores[key]

        mean = np.mean(values)

        sd = np.std(
            values,
            ddof=1
        )

        se = (
            sd /
            np.sqrt(len(values))
        )

        lower = (
            mean
            -
            1.96 * se
        )

        upper = (
            mean
            +
            1.96 * se
        )

        ci_rows.append({

            "Model":
                model_name,

            "Metric":
                metric_name,

            "Mean":
                mean,

            "SD":
                sd,

            "95% CI Lower":
                lower,

            "95% CI Upper":
                upper
        })


ci_table = pd.DataFrame(
    ci_rows
)


print(
    ci_table
    .round(4)
    .to_string(index=False)
)


# 10. BOXPLOT — ACCURACY DISTRIBUTION

plt.figure(figsize=(8, 6))

plt.boxplot(
    [
        original_scores["test_accuracy"],
        proposed_scores["test_accuracy"]
    ],
    labels=[
        "Original Stacking",
        "Proposed Tuned Stacking"
    ]
)

plt.ylabel("Accuracy")

plt.title(
    "Repeated Cross-Validation Accuracy\n"
    "Deduplicated Dataset (50 Test Folds)"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/10_repeated_cv_accuracy_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# 11. BOXPLOT — AUC DISTRIBUTION

plt.figure(figsize=(8, 6))

plt.boxplot(
    [
        original_scores["test_auc"],
        proposed_scores["test_auc"]
    ],
    labels=[
        "Original Stacking",
        "Proposed Tuned Stacking"
    ]
)

plt.ylabel("ROC-AUC")

plt.title(
    "Repeated Cross-Validation AUC\n"
    "Deduplicated Dataset (50 Test Folds)"
)

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/11_repeated_cv_auc_boxplot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# 12. SAVE RESULTS

robust_table.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_summary.csv",
    index=False
)

improvement_table.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_improvement.csv",
    index=False
)

stats_table.to_csv(
    f"{OUTPUT_DIR}/wilcoxon_statistical_tests.csv",
    index=False
)

ci_table.to_csv(
    f"{OUTPUT_DIR}/repeated_cv_confidence_intervals.csv",
    index=False
)

# 13. FINAL COMPACT SUMMARY

print("FINAL STATISTICAL VALIDATION SUMMARY")

print(
    "\nValidation design:"
)

print(
    "5-fold stratified CV × 10 repetitions = 50 paired test folds"
)

print(
    "Dataset: 302 unique observations after exact duplicate removal"
)


print("\nORIGINAL STACKING")

print(
    f"Accuracy = "
    f"{np.mean(original_scores['test_accuracy']):.4f} "
    f"± {np.std(original_scores['test_accuracy']):.4f}"
)

print(
    f"F1       = "
    f"{np.mean(original_scores['test_f1']):.4f} "
    f"± {np.std(original_scores['test_f1']):.4f}"
)

print(
    f"AUC      = "
    f"{np.mean(original_scores['test_auc']):.4f} "
    f"± {np.std(original_scores['test_auc']):.4f}"
)


print("\nPROPOSED TUNED STACKING")

print(
    f"Accuracy = "
    f"{np.mean(proposed_scores['test_accuracy']):.4f} "
    f"± {np.std(proposed_scores['test_accuracy']):.4f}"
)

print(
    f"F1       = "
    f"{np.mean(proposed_scores['test_f1']):.4f} "
    f"± {np.std(proposed_scores['test_f1']):.4f}"
)

print(
    f"AUC      = "
    f"{np.mean(proposed_scores['test_auc']):.4f} "
    f"± {np.std(proposed_scores['test_auc']):.4f}"
)


print("\nSTATISTICAL TESTS")

for _, row in stats_table.iterrows():

    print(
        f"{row['Metric']:10s}: "
        f"p = {row['p-value']:.6f} "
        f"({row['Significant at 0.05']})"
    )


print("\nSaved files:")

print("• repeated_cv_summary.csv")
print("• repeated_cv_improvement.csv")
print("• wilcoxon_statistical_tests.csv")
print("• repeated_cv_confidence_intervals.csv")
print("• 10_repeated_cv_accuracy_boxplot.png")
print("• 11_repeated_cv_auc_boxplot.png")

print("STATISTICAL VALIDATION COMPLETE")
```
